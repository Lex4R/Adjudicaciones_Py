# Hipotesis
## Formulacion en lenguaje natural:

#### * H0: No existe asociacion ni dependencia entre la variable **categoria_principal** y la variable de presencia de ampliacion: **target_ampliacion_monto**
#### * H1: Existe asociacion estadistica significativa entre la categoria principal y la presencia de ampliacion monetaria

## Formulacion en lenguaje Formal

#### * H0: P(Categorias[i] n Ampliacion[j]) = P(C) * P(A) para todas las categorias i que pertenecen{goods, works, services} y estados de ampliación j en {0, 1}.
#### * H1: P(Categorias[i] n Ampliacion[j]) != P(C) * P(A) para al menos un par (i , j)

# Definicion de las variables y tabla de contingencia

```python
datos = df.copy()
datos['target_ampliacion_monto'] = datos['target_ampliacion_monto'].astype(int)
```

```python
tabla_contingencia = pd.crosstab(datos['categoria_principal'], datos['target_ampliacion_monto'])
print("Mostrar tabla de contingencia:")
display(tabla_contingencia)
```

## Ejecucion de chi-cuadrado y verificacion de supuestos

```python
import scipy.stats as stats
chi2, p, dof, esperadas = stats.chi2_contingency(tabla_contingencia)
print("Chi-cuadrado:", chi2)
print("Valor p:", p)
print("Grados de libertad:", dof)
print("Frecuencias esperadas:")
display(esperadas)

print("Frecuencia esperada con menor valor:", np.min(esperadas))
print("Frecuencia esperada menor a 5:", bool((esperadas < 5).sum()))

```

## Calculo del tamanho del efecto V de cramer

```python
n = tabla_contingencia.to_numpy().sum()
filas, columnas = tabla_contingencia.shape
phi2 = chi2 / n
v_crammer = np.sqrt(phi2 / min(filas-1, columnas-1))
print("Número total de observaciones:", n)
print("Número de filas:", filas)
print("Número de columnas:", columnas)
print("Phi-cuadrado:", phi2)
print("V de Cramér:", v_crammer)
```

# Conclusion Final:
##### Podemos afirmar con el valor de p-valor obtenido (p < 0.0001) que existe evidencia estadistica suficiente para rechazar la hipotesis nula y determinar que existe relacion entre la variable categoria_principal y target_ampliacion_monto.<br> Por su parte, V de cramer(0.1675) permite afirmar que existe una fuerza de asociacion entre las variables de debil a moderada.<br> Es importante aclarar, que los resultados demuestran la dependencia entre las variables no asi la causalidad entre ellas.
